```python
#Fatma, Youssra, Lyes, Thomas

sales = [120, 95, 140, 180, 75, 160]

# 1. Number of orders, total and mean

nombre = 0
total=0
for i in sales :
    nombre+=1
    total+=i
mean = total / nombre
print("Nombre :",nombre,"Total : ",total," Moyenne ",mean)

# 2. Define classify_sales(value), then apply it to each amount
print("")
def classify_sales(val):
    if val>150:
        return "High"
    return "Standard"

for i in sales :
    print(classify_sales(i))

# 3. Build the list of high amounts using a loop and a condition
print("")
list = []
for i in sales:
    if classify_sales(i)=="High":
        list.append(i)
print(list)

# 4. Build and display report
print("")
report = {
    "nombre":nombre,
    "total":total,
    "moyenne":mean,
}
print(report)
```
